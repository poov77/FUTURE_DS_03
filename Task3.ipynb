```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')

# 1. LOAD DATA
csv_path = r'K:\future Intern\Task 3\bank-full.csv'

# bank-full.csv in this folder is comma-separated.
df = pd.read_csv(csv_path)
df.columns = df.columns.str.strip().str.strip('"').str.lower()

if 'y' not in df.columns:
    raise KeyError(f"Column 'y' was not found. Available columns are: {df.columns.tolist()}")

# 2. DATA PREPARATION
df['conversion_flag'] = (df['y'].astype(str).str.strip().str.lower() == 'yes').astype(int)
df['campaign_stage'] = pd.cut(
    df['campaign'],
    bins=[0, 1, 2, 3, 5, float('inf')],
    labels=['1st contact', '2nd contact', '3rd contact', '4-5 contacts', '6+ contacts']
)
df['duration_bucket'] = pd.cut(
    df['duration'],
    bins=[0, 60, 180, 300, 600, float('inf')],
    labels=['under 1 min', '1-3 min', '3-5 min', '5-10 min', '10+ min']
)

overall_conversion = df['conversion_flag'].mean() * 100
print(f"Overall conversion rate: {overall_conversion:.2f}%")

# 3. WHERE ARE USERS DROPPING OFF IN THE FUNNEL?
funnel_dropoff = df.groupby('campaign_stage', observed=False).agg(
    leads=('conversion_flag', 'size'),
    conversions=('conversion_flag', 'sum'),
    conversion_rate=('conversion_flag', 'mean')
).reset_index()
funnel_dropoff['conversion_rate'] *= 100
funnel_dropoff['lead_share'] = funnel_dropoff['leads'] / len(df) * 100
funnel_dropoff['drop_from_previous_stage'] = funnel_dropoff['conversion_rate'].diff()

print("\nWhere are users dropping off?")
display(funnel_dropoff.round(2))

# 4. WHICH CHANNELS BRING HIGH-QUALITY LEADS?
channel_quality = df.groupby('contact').agg(
    leads=('conversion_flag', 'size'),
    conversions=('conversion_flag', 'sum'),
    conversion_rate=('conversion_flag', 'mean'),
    avg_call_duration=('duration', 'mean'),
    avg_contacts=('campaign', 'mean')
).reset_index()
channel_quality['conversion_rate'] *= 100
channel_quality = channel_quality.sort_values('conversion_rate', ascending=False)

print("\nWhich channels bring high-quality leads?")
display(channel_quality.round(2))

# 5. HOW CAN CONVERSION RATES BE IMPROVED?
segment_opportunity = df.groupby(['contact', 'poutcome', 'housing']).agg(
    leads=('conversion_flag', 'size'),
    conversions=('conversion_flag', 'sum'),
    conversion_rate=('conversion_flag', 'mean')
).reset_index()
segment_opportunity['conversion_rate'] *= 100
segment_opportunity = segment_opportunity[segment_opportunity['leads'] >= 100].sort_values('conversion_rate', ascending=False)

print("\nBest lead segments to prioritize for conversion improvement:")
display(segment_opportunity.head(10).round(2))

# 6. WHICH STAGES NEED OPTIMIZATION?
stage_optimization = df.groupby(['campaign_stage', 'duration_bucket'], observed=False).agg(
    leads=('conversion_flag', 'size'),
    conversion_rate=('conversion_flag', 'mean')
).reset_index()
stage_optimization['conversion_rate'] *= 100
stage_optimization = stage_optimization[stage_optimization['leads'] >= 50]

print("\nStages needing optimization: look for high lead volume with low conversion rate.")
display(stage_optimization.sort_values(['conversion_rate', 'leads'], ascending=[True, False]).head(10).round(2))

# 7. VISUALIZATIONS
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.lineplot(data=funnel_dropoff, x='campaign_stage', y='conversion_rate', marker='o', ax=axes[0])
axes[0].set_title('Drop-off by Contact Stage')
axes[0].set_xlabel('Campaign Stage')
axes[0].set_ylabel('Conversion Rate (%)')
axes[0].tick_params(axis='x', rotation=30)

sns.barplot(data=channel_quality, x='contact', y='conversion_rate', ax=axes[1])
axes[1].set_title('Lead Quality by Channel')
axes[1].set_xlabel('Channel')
axes[1].set_ylabel('Conversion Rate (%)')

top_segments = segment_opportunity.head(8).copy()
top_segments['segment'] = top_segments['contact'] + ' | ' + top_segments['poutcome'] + ' | housing: ' + top_segments['housing']
sns.barplot(data=top_segments, y='segment', x='conversion_rate', ax=axes[2])
axes[2].set_title('Top Lead Segments')
axes[2].set_xlabel('Conversion Rate (%)')
axes[2].set_ylabel('')

plt.tight_layout()
plt.show()
```
